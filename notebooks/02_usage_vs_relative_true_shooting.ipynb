# Usage vs. Relative True Shooting by Position

x-axis: usage %. y-axis: TS% minus the position (or position group) average, in percentage points. Dots are colored by team (traded players use their last team). Players with fewer than 20 games or under 6 minutes per game are excluded. Positions are Basketball-Reference's listed position; groups are G = PG + SG, F = SF + PF, C. Only the most extreme points are labeled. Dashed lines mark the averages. Hover a dot for the player's details; click a team in the legend to hide it, or double-click to isolate it. Charts are interactive in JupyterLab or VS Code; GitHub shows a static image.

Source: Basketball-Reference.

```python
import pandas as pd
import plotly.io as pio

from nba_analytics.config import POSITIONS, season_label
from nba_analytics.players.usage_efficiency import plot_usage_ts_interactive, usage_ts_frame

# Interactive in JupyterLab / VS Code; static PNG fallback where Plotly can't run (e.g. GitHub)
pio.renderers.default = "plotly_mimetype+png"
pio.renderers["png"].scale = 0.7  # keeps the notebook small enough for GitHub to render

SEASONS = [2023, 2024, 2025, 2026]
frames = {s: usage_ts_frame(s) for s in SEASONS}
```

```python
pd.DataFrame({season_label(s): df.groupby('pos').size() for s, df in frames.items()}).rename_axis('Qualifying players')
```

## 2025-26

### Positions

```python
for pos in POSITIONS:
    plot_usage_ts_interactive(frames[2026], pos, relative=True).show()
```

### Position groups

```python
for group in ['G', 'F', 'C']:
    plot_usage_ts_interactive(frames[2026], group, relative=True).show()
```
