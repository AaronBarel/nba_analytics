# Usage vs. True Shooting by Position

x-axis: usage %. y-axis: true shooting %. Dots are colored by team (traded players use their last team). Players with fewer than 20 games or under 6 minutes per game are excluded. Positions are Basketball-Reference's listed position; groups are G = PG + SG, F = SF + PF, C. Only the most extreme points are labeled. Dashed lines mark the averages.

Source: Basketball-Reference.

```python
import warnings

import matplotlib.pyplot as plt
import pandas as pd

from nba_analytics.config import POSITIONS, season_label
from nba_analytics.players.usage_efficiency import plot_usage_ts, usage_ts_frame

warnings.filterwarnings("ignore", module="adjustText")
%matplotlib inline
plt.rcParams["figure.dpi"] = 80

SEASONS = [2023, 2024, 2025, 2026]
frames = {s: usage_ts_frame(s) for s in SEASONS}
```

```python
pd.DataFrame({season_label(s): df.groupby('pos').size() for s, df in frames.items()}).rename_axis('Qualifying players')
```

## 2025-26

### Positions

```python
for pos in POSITIONS:
    plot_usage_ts(frames[2026], pos, relative=False)
    plt.show()
```

### Position groups

```python
for group in ['G', 'F', 'C']:
    plot_usage_ts(frames[2026], group, relative=False)
    plt.show()
```
